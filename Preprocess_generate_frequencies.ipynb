# Generate `English_dict_frequencies.txt`

Appendix E (`Appendix_E_generate_Wk_Sk.ipynb`) expects two files in the same folder,
aligned line-by-line:

- `English_dict_words.txt` -- the word list (rename/point this at `words_alpha.txt`
  from https://github.com/dwyl/english-words)
- `English_dict_frequencies.txt` -- each word's `wordfreq` frequency, **in the same order**

This notebook builds the second file from the first. It's a one-off preprocessing
step (roughly 466k words), so it's kept separate from Appendix E rather than run
every time you open that notebook.

**Run this once**, then reuse the resulting `English_dict_frequencies.txt` going forward.


```python
# Run once if wordfreq / tqdm aren't already installed
# %pip install wordfreq tqdm
```

```python
import numpy as np
from tqdm import tqdm
from wordfreq import word_frequency

# --- Config ---
WORDS_FILE = 'words_alpha.txt'                  # input: one word per line
FREQ_FILE = 'English_dict_frequencies.txt'       # output: frequency per line, same order
WORDS_OUT_FILE = 'English_dict_words.txt'        # optional: a renamed copy matching Appendix E's expected filename
LANG = 'en'

# --- Load words ---
words = np.genfromtxt(WORDS_FILE, delimiter='\n', dtype=str)
print(f"Loaded {len(words)} words from {WORDS_FILE}")

# --- Compute frequencies (this is the slow part) ---
freqs = [word_frequency(w, LANG) for w in tqdm(words, desc="Computing word frequencies")]

# --- Save frequencies, aligned line-by-line with `words` ---
np.savetxt(FREQ_FILE, freqs, fmt='%.10g')
print(f"Wrote {len(freqs)} frequencies to {FREQ_FILE}")

# --- Optional: write a copy of the word list under the filename Appendix E expects,
#     so you don't have to edit paths there. Skip this if you'd rather just edit
#     Appendix E to read 'words_alpha.txt' directly. ---
np.savetxt(WORDS_OUT_FILE, words, fmt='%s')
print(f"Wrote a copy of the word list to {WORDS_OUT_FILE}")
```

## Sanity check

A quick look at the most and least frequent words, and confirming word count.


```python
freqs_arr = np.array(freqs)
order = np.argsort(freqs_arr)[::-1]

print("Most frequent words:")
for i in order[:10]:
    print(f"  {words[i]:<15} {freqs_arr[i]:.6g}")

print("\nWords with frequency 0 (rare/unknown to wordfreq):", int(np.sum(freqs_arr == 0.0)))
print("Total words:", len(words), "| Total frequencies:", len(freqs_arr))
```
